```python
import pandas as pd
import sqlite3
import numpy as np 

DB_PATH = '../backend/data.db'

conn = sqlite3.Connection(DB_PATH)
```

```python
# Top 10%
pd.read_sql("""
    select article_id
    from articles 
    where count > 150 and 
        (length(title) - length(replace(title, ' ', '')))
            / nullif(length(' '), 0) = 0
    order by random() 
    limit 1
            
""", conn)
```
